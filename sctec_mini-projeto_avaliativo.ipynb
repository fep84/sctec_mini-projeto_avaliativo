## E) - Ajustar campo ‘quantidade’ (inteiro e maior que zero

```python
categoria_texto = df["categoria"].astype("string").str.strip()
categorias_vazias = categoria_texto.isna() | categoria_texto.eq("").fillna(False)

print("Categorias vazias:", int(categorias_vazias.sum()))

df["categoria"] = categoria_texto.mask(categorias_vazias, "sem categoria")
```
