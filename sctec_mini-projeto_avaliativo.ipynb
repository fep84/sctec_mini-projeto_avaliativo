```python
import pandas as pd

df = pd.read_csv("halloween_vendas.csv")
```

#### 1.3) - Confira as linhas, os nomes das colunas e o total de registros.

```python
# len(df) conta as linhas, e df.columns.tolist() mostra os nomes das colunas. 
# Confere se elas se chamam exatamente categoria, quantidade e preco_unitario.

print("Total de linhas:", len(df))
print("Nomes das colunas:", df.columns.tolist())
display(df.head())
```

### 1.4) - Preencher categorias vazias com a string "sem categoria".

```python
# Conta categorias vazias antes de preenchê-las
categoria_texto = df["categoria"].astype("string").str.strip()
categorias_vazias = categoria_texto.isna() | categoria_texto.eq("").fillna(False)


print("Categorias vazias:", int(categorias_vazias.sum()))


# Preenche sem alterar as demais categorias
df["categoria"] = categoria_texto.mask(categorias_vazias, "sem categoria")
```

### 1.5) - Verificar se quantidade é inteira e maior que zero
### 1.6) - Verificar se preco_unitario está preenchido e não é negativo
### 1.7) - Marcar os registros com problemas


```python
# pd.to_numeric(..., errors="coerce") tenta interpretar os valores como números; 
# valores vazios ou inválidos ficam marcados como ausentes nas variáveis usadas para a verificação. 
# As colunas originais não são corrigidas nem substituídas.

quantidade_num = pd.to_numeric(df["quantidade"], errors="coerce")
quantidade_valida = (
    quantidade_num.notna()
    & quantidade_num.gt(0)
    & quantidade_num.mod(1).eq(0)
)
df["problema_quantidade"] = ~quantidade_valida


preco_num = pd.to_numeric(df["preco_unitario"], errors="coerce")
preco_valido = preco_num.notna() & preco_num.ge(0)
df["problema_preco_unitario"] = ~preco_valido


# Marca a linha se houver problema em quantidade ou preço.
df["tem_problema_numerico"] = (
    df["problema_quantidade"] | df["problema_preco_unitario"]
)

```
